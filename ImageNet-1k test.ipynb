```python
from datasets import load_dataset
import matplotlib.pyplot as plt
import random

# Load the validation split of the ImageNet dataset
dataset = load_dataset("imagenet-1k", split="validation", trust_remote_code=True)

# Check the structure of the dataset
print(f"Dataset contains {len(dataset)} samples.")

# Check the class labels available
label_info = dataset.info.features['label']
num_classes = label_info.num_classes
label_names = label_info.names

print(f"Dataset contains {num_classes} classes.")
print(f"Sample class labels: {label_names[:10]}")  # Adjust this number to see more class names

# Function to display a specified number of random samples in a grid
def show_random_samples(num_samples=25):
    # Determine the number of rows and columns for the subplot grid
    num_columns = 5
    num_rows = (num_samples + num_columns - 1) // num_columns  # This ensures we have enough rows

    # Create a figure with appropriate size
    fig, axes = plt.subplots(num_rows, num_columns, figsize=(15, num_rows * 3))

    # Flatten axes array for easy iteration, if there are multiple rows and columns
    axes = axes.flatten()

    # Plot each sample
    for idx in range(num_samples):
        ax = axes[idx]
        sample = random.choice(dataset)
        image = sample['image']
        label = label_names[sample['label']]

        # Display the image and label
        ax.imshow(image)
        ax.set_title(f'Label: {label}', wrap=True)
        ax.axis('off')

    # Turn off axes for any unused subplots
    for idx in range(num_samples, len(axes)):
        axes[idx].axis('off')

    plt.tight_layout()
    plt.show()

# Display 25 random samples
show_random_samples(25)  # You can change this number to another reasonable value like 50 or 100
```
