````python
import numpy as np
from scipy import ndimage
from scipy.ndimage import center_of_mass
from sklearn.cluster import KMeans

class HeatmapCenter:
    def __init__(self, threshold=None):
        self.threshold = threshold

    def __call__(self, heatmap):
        if self.threshold is None:
            self.threshold = np.mean(heatmap) 
        mask = heatmap > self.threshold

        center = ndimage.center_of_mass(mask)

        return np.array([[int(center[1]), int(center[0])]]), np.array([1])

class HeatmapMultipleCenters:
    def __init__(self, threshold=None, min_size=10):
        self.threshold = threshold
        self.min_size = min_size

    def __call__(self, heatmap):
        if self.threshold is None:
            self.threshold = np.mean(heatmap) 
        mask = heatmap > self.threshold

        labeled, num_features = ndimage.label(mask)
        
        centers = []
        for i in range(1, num_features + 1):
            component = (labeled == i)
            if np.sum(component) >= self.min_size:
                center = ndimage.center_of_mass(component)
                centers.append((int(center[1]), int(center[0])))

        if len(centers) == 0:
            return HeatmapCenter(self.threshold)(heatmap)
        
        return np.array(centers), np.ones(len(centers))

class PeakIntensity:
    def __init__(self, num_peaks=1):
        self.num_peaks = num_peaks

    def __call__(self, heatmap):
        flat_indices = np.argpartition(heatmap.ravel(), -self.num_peaks)[-self.num_peaks:]
        
        return np.array([np.unravel_index(idx, heatmap.shape)[::-1] for idx in flat_indices]), np.ones(self.num_peaks)

class Centroid:
    def __call__(self, heatmap):
        total_mass = np.sum(heatmap)
        h, w = heatmap.shape
        y_coords, x_coords = np.mgrid[:h, :w]
        cx = np.sum(x_coords * heatmap) / total_mass
        cy = np.sum(y_coords * heatmap) / total_mass
        return np.array([[int(cx), int(cy)]]), np.array([1])

class MedianPoint:
    def __init__(self, threshold=None):
        self.threshold = threshold

    def __call__(self, heatmap):
        if self.threshold is None:
            self.threshold = np.median(heatmap)
        
        y_coords, x_coords = np.where(heatmap > self.threshold)
        if len(y_coords) == 0:
            return None  # No points above threshold
        return np.array([[int(np.median(x_coords)), int(np.median(y_coords))]]), np.array([1])

class MultipleCentersThreshold:
    def __init__(self, threshold=None):
        self.threshold = threshold

    def __call__(self, heatmap):
        if self.threshold is None:
            self.threshold = np.mean(heatmap) + np.std(heatmap)
        
        binary = heatmap > self.threshold
        labeled, num_features = ndimage.label(binary)
        centers = center_of_mass(heatmap, labeled, range(1, num_features + 1))

        if len(centers) == 0:
            return HeatmapCenter(self.threshold)(heatmap)
        
        return np.array([(int(x), int(y)) for y, x in centers]), np.ones(num_features)

class MultipleCentersKMeans:
    def __init__(self, n_clusters=3):
        self.n_clusters = n_clusters

    def __call__(self, heatmap):
        y_coords, x_coords = np.nonzero(heatmap)
        points = np.column_stack((y_coords, x_coords))
        weights = heatmap[y_coords, x_coords]
        kmeans = KMeans(n_clusters=self.n_clusters)
        kmeans.fit(points, sample_weight=weights)
        return np.array([(int(center[1]), int(center[0])) for center in kmeans.cluster_centers_]), np.ones(self.n_clusters)

class GetMaxPoint:
    def __call__(self, heatmap):
        idx = np.unravel_index(np.argmax(heatmap, axis=None), heatmap.shape)
        return np.array([[idx[1], idx[0]]], dtype=int), np.array([1])

class SampleNPoints:
    def __init__(self, n_samples, power=1):
        self.n_samples= n_samples
        self.power = power

    def __call__(self, heatmap):
        flat_heatmap = heatmap.flatten()
        flat_heatmap = flat_heatmap ** self.power
        prob_dist = flat_heatmap / np.sum(flat_heatmap)
        
        # Sample indices based on the probability distribution
        sampled_indices = np.random.choice(
            len(flat_heatmap), 
            size=self.n_samples, 
            replace=False,  # Sample without replacement
            p=prob_dist
        )
        
        # Convert flat indices back to 2D coordinates
        sampled_points = [np.unravel_index(idx, heatmap.shape) for idx in sampled_indices]
        
        return np.array(sampled_points), np.ones(self.n_samples)
    
def sample_points_histogram(heatmap, n_samples, n_bins=10):
    """
    Sample N points from the heatmap based on a histogram of intensities.
    
    Args:
    heatmap (numpy.ndarray): 2D array representing the heatmap
    n_samples (int): Number of points to sample
    n_bins (int): Number of bins for the histogram
    
    Returns:
    list of tuples: Sampled points as (row, col)
    """
    # Flatten the heatmap
    flat_heatmap = heatmap.flatten()
    
    # Create histogram
    hist, bin_edges = np.histogram(flat_heatmap, bins=n_bins)
    
    # get samples from first N bins
    sampled_indices = []
    for i in range(n_bins):
        indices = np.where((flat_heatmap >= bin_edges[i]) & (flat_heatmap < bin_edges[i + 1]))[0]
        if len(indices) > 0:
            sampled_indices.extend(np.random.choice(indices, n_samples // n_bins, replace=False))

    # Convert flat indices back to 2D coordinates
    sampled_points = [np.unravel_index(idx, heatmap.shape) for idx in sampled_indices]

    return np.array(sampled_points), np.ones(n_samples)


point_methods = {
    "HeatmapCenter": HeatmapCenter(),
    "HeatmapMultipleCenters": HeatmapMultipleCenters(),
    "PeakIntensity": PeakIntensity(),
    "Centroid": Centroid(),
    "MedianPoint": MedianPoint(),
    "MultipleCentersThreshold": MultipleCentersThreshold(),
    "MultipleCentersKMeans": MultipleCentersKMeans(),
    "GetMaxPoint": GetMaxPoint(),
    "SampleNPoints": SampleNPoints(3),
}
````

````python

def show_points(coords, labels, ax, marker_size=375):
    pos_points = coords[labels==1]
    neg_points = coords[labels==0]
    ax.scatter(pos_points[:, 0], pos_points[:, 1], color='green', marker='*', s=marker_size, edgecolor='white', linewidth=1.25)
    ax.scatter(neg_points[:, 0], neg_points[:, 1], color='red', marker='*', s=marker_size, edgecolor='white', linewidth=1.25)   
    
def show_mask(mask, ax, random_color=False):
    if random_color:
        color = np.concatenate([np.random.random(3), np.array([0.6])], axis=0)
    else:
        color = np.array([30/255, 144/255, 255/255, 0.6])
    h, w = mask.shape[-2:]
    mask_image = mask.reshape(h, w, 1) * color.reshape(1, 1, -1)
    ax.imshow(mask_image)

````

Extract labels from segmentation masks (see [here](https://github.com/LUSSeg/ImageNet-S?tab=readme-ov-file#qa) for more details).

````python
def get_seg_id(path):
    segmentation = Image.open(path)  # RGB
    segmentation = np.array(segmentation, dtype=np.uint16)
    seg = segmentation[:, :, 1] * 256 + segmentation[:, :, 0]  # R+G*256
    seg = np.unique(seg)
    seg = seg[(seg != 0) & (seg != 1000)]
    return seg


imgnet_s_df["segmentation_id"] = imgnet_s_df["file_path"].apply(lambda x: get_seg_id(x))
imgnet_s_df.head()
````
